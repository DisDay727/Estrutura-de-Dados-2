```python
class Simulador:
    def __init__(self):
        self.texto = []

    def adiciona(self, x):
        self.texto.append(x)
        for i in range(len(self.texto)):
            print(f"{i} Nome: {self.texto[i]}")
        print(f"O User {x} foi adicionado")
        print("")

    def desfazer(self):
        if self.texto:
            self.texto.pop()
        else:
            print("Lista vazia, nada para desfazer.")

    def apagar(self,x):
        found = False
        for i in range(len(self.texto) - 1, -1, -1):
            if self.texto[i] == x:
                print(f"Usuario {self.texto[i]} foi apagado")
                self.texto.pop(i)
                found = True
                break
        if not found:
            print("Usuario não encontrado")

    def edita(self,x):
        found = False
        for i in range(len(self.texto)):
            if self.texto[i] == x:
                print(f"O nome {self.texto[i]} foi encontrado")
                aux=input("qual é o novo nome ? ")
                self.texto[i]=aux
                found = True
                break
        if not found:
            print("Usuario não encontrado")

    def lista(self):
      if not self.texto:
          print("A lista está vazia.")
          return
      for i in range(len(self.texto)):
        print(f"indice:{i} Nome:{self.texto[i]}")
simulador = Simulador()
```

# Adiciona User

```python

simulador.adiciona("Max")
simulador.adiciona("Lisa")
simulador.adiciona("Bart")
print("")
simulador.lista()
print("")
```

#Apaga usando pop

```python
simulador.apagar("Max")
simulador.lista()
print("")
```

#Adiciona Max e desfaz

```python
simulador.adiciona("Max")
simulador.lista()
print("#Adicionou Max#")
simulador.desfazer()
print("Desfez")
simulador.lista()
```

#Edita por nome

```python
simulador.edita("Lisa")
print("")
simulador.lista()
```
